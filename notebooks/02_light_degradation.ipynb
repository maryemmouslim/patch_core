# Degradation photometrique

Compare la baseline sous reference, faible, fort, bleu et jaune.

```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
while project_root != project_root.parent:
    if (project_root / 'src').is_dir() and (project_root / 'mvtec_anomaly_detection').is_dir():
        break
    project_root = project_root.parent
else:
    raise RuntimeError('Racine du projet LightShift-PatchCore introuvable')

%cd {project_root}
sys.path.insert(0, str(project_root))
```

```python
import subprocess
import sys

for lighting in ['reference', 'dim', 'bright', 'blue', 'yellow']:
    subprocess.run([
        sys.executable,
        '-m', 'src.experiments.run_experiments',
        '--root', 'mvtec_anomaly_detection',
        '--categories', 'bottle',
        '--lighting', lighting,
        '--batch-size', '4'
    ], cwd=project_root, check=True)
```

```python
import glob
import json
import pandas as pd

rows = []
for path in glob.glob(str(project_root / 'artifacts/bottle/*/metrics.json')):
    with open(path, encoding='utf-8') as handle:
        rows.append(json.load(handle))

pd.DataFrame(rows).sort_values('lighting')[['lighting', 'image_auroc', 'pixel_auroc', 'threshold', 'accuracy']]
```
