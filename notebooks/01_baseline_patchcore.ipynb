# Baseline PatchCore

Ce notebook lance la baseline MVTec sur `bottle` et affiche les métriques.

```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
while project_root != project_root.parent:
    if (project_root / 'src').is_dir() and (project_root / 'mvtec_anomaly_detection').is_dir():
        break
    project_root = project_root.parent
else:
    raise RuntimeError('Racine du projet LightShift-PatchCore introuvable')

%cd {project_root}
sys.path.insert(0, str(project_root))
print(project_root)
```

```python
import subprocess
import sys

metrics_path = project_root / 'artifacts/bottle/reference/metrics.json'
if not metrics_path.exists():
    subprocess.run([
        sys.executable,
        '-m', 'src.experiments.run_experiments',
        '--root', 'mvtec_anomaly_detection',
        '--categories', 'bottle',
        '--lighting', 'reference',
        '--batch-size', '4'
    ], cwd=project_root, check=True)
else:
    print('Baseline deja disponible:', metrics_path)
```

```python
import json

with open(project_root / 'artifacts/bottle/reference/metrics.json', encoding='utf-8') as handle:
    print(json.dumps(json.load(handle), indent=2))
```
